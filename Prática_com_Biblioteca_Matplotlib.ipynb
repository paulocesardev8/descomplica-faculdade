```python
import numpy as np
import matplotlib.pyplot as plt

# 2. Criação do conjunto de dados
x = np.linspace(0, 10, 100)  # 100 valores de 0 a 10
y1 = np.sin(x)  # Valores de seno
y2 = np.cos(x)  # Valores de cosseno

# 3. Criando um gráfico de linha
plt.figure(figsize=(8, 5))
plt.plot(x, y1, label='Seno', color='blue')
plt.title('Gráfico de Linha - Função Seno')
plt.xlabel('X')
plt.ylabel('Seno(X)')
plt.legend()
plt.show()

# 4. Criando um gráfico de dispersão
plt.figure(figsize=(8, 5))
plt.scatter(x, y2, label='Cosseno', color='red')
plt.title('Gráfico de Dispersão - Função Cosseno')
plt.xlabel('X')
plt.ylabel('Cosseno(X)')
plt.legend()
plt.show()

# 5. Criando um gráfico de barras
valores = [3, 7, 1, 5, 9]  # Valores inteiros
rotulos = ['A', 'B', 'C', 'D', 'E']  # Rótulos para as barras
plt.figure(figsize=(8, 5))
plt.bar(rotulos, valores, color='green')
plt.title('Gráfico de Barras')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.savefig('grafico_barras.png')  # 7. Salvando o gráfico
plt.show()

# 6. Criando subplots
fig, axs = plt.subplots(2, 1, figsize=(8, 10))

# Gráfico de Linha no primeiro subplot
axs[0].plot(x, y1, label='Seno', color='blue')
axs[0].set_title('Subplot 1 - Gráfico de Linha')
axs[0].set_xlabel('X')
axs[0].set_ylabel('Seno(X)')
axs[0].legend()

# Gráfico de Dispersão no segundo subplot
axs[1].scatter(x, y2, label='Cosseno', color='red')
axs[1].set_title('Subplot 2 - Gráfico de Dispersão')
axs[1].set_xlabel('X')
axs[1].set_ylabel('Cosseno(X)')
axs[1].legend()

plt.tight_layout()
plt.show()
```
